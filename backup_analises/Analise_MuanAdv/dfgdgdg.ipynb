```python
#
import numpy as np
import matplotlib.pyplot as plt 
import math 
import rpcsimprog as rpc
from scipy import integrate 
from scipy.stats import norm 
from scipy import stats
#
#  Read Experimental Data into a Pads matrix
#
dir = "/Volumes/RCS2_128GB/rpc_simulation/data/"
direxp = "/Volumes/RCS2_128GB/rpc_data/"
fig = plt.figure()
#
#alphas = (1.25, 1.44, 1.63, 1.78, 1.95, 2.13)
index = (8, 9)
#
for alpha_exp in index:
    ndata = 30
    alphax = np.zeros(ndata)
    chi2x  = np.zeros(ndata)
    file = direxp + "R0014_MAROC_" + str(alpha_exp) + "_06102016.bin"
    print("Will read experimentaldata from file: ", file)
    matrixExp = rpc.ReadDataExp(file)

    alpha = 0.5
    for i in range(0,ndata):
        alpha += 0.10
        filesim = dir + "alpha_" + str("{0:4.2f}".format(alpha)) + "_Pad_sim.txt"
        matrixPads = rpc.ReadDataExp(filesim)
        chi2 = rpc.Chi2Matrices(matrixPads, matrixExp)
        alphax[i] = alpha
        chi2x[i]  = chi2                
#
        print("{0:12s}{1:3.1f}{2:11s}{3:8.2f}".format("For alpha = ", alpha, " -> Chi2 = ", chi2))
#
    figpdf = 'Plot_Chi2_' + str(alpha_exp) + '.pdf'
    print ("Plot for alpha = : ", alpha_exp, " file: ", figpdf)

#
    fig = plt.figure()
    plt.plot(alphax, chi2x, linewidth = 2.0, color='red', marker='o')
    plt.ylabel("Chi2")
    plt.xlabel("alpha_exp: " + str(alpha_exp))
    plt.savefig(figpdf)
    plt.close(fig)
#    
print("-- Finished job")
```
